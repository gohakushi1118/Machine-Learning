```python
import torch
import torch.nn as nn
import numpy as np
```

```python
x = torch.empty(5, 3) # 空張量
print(x)
```

```python
x = torch.rand(5, 3) # 隨機張量
print(x)
```

```python
x = torch.zeros(5, 3, dtype=torch.int) # 零張量
print(x)
```

```python
x = torch.tensor([[1, 2], [3, 4]]) # 多維矩陣
print(x)
```

```python
x = torch.ones(5, 3, dtype=torch.double) # 單位張量
print(x)
```

```python
z = torch.empty(5, 3)
# 創建新同為度的張量
x = torch.rand_like(z, dtype=torch.double) 
y = torch.ones_like(z, dtype=torch.double)
print(x, "\n")
print(y)
```

```python
x = torch.ones(5, 3) # 張量大小
print(x.size())
```

```python
x = torch.ones(5, 3)
y = torch.ones(5, 3)
# 張量運算
print(x+y)
print(x-y)
print(x*y)
print(x**2)
```

```python
x = torch.tensor([[1, 2], [3, 4]])
print(x[:, 0]) # 第一行
print(x[:, 1]) # 第二行
print(x[0, :]) # 第一列
print(x[1, :]) # 第二列
```

```python
x = torch.rand(4, 4)
y = x.view(16) # 多維張量攤平為一維張量
print(y)
```

```python
x = torch.rand(4, 4)
y = x.view(2, 8) # 張量大小修飾
print(y)
```

```python
x = torch.rand(1)
print(x.item())
```

```python
x = torch.rand(3, 3) # torch to numpy
y = x.numpy()
print(y, "\n")
print(type(y))
```

```python
x = torch.rand(3, 3, requires_grad=True) # 自動微分系統
print(x)
```

```python
x = torch.rand(3, 3, requires_grad=True)
y = x + 2
print(y.grad_fn) # 用來記錄這個張量由哪個運算產生的

y = x * 3
print(y.grad_fn)
```

```python
x = torch.rand(3, 3, requires_grad=True)
y = x*2
y = y.mean()

print(x)
print(y)

y.backward() # 計算微分後的結果
print(x.grad) # 將 y.backward 結果存入 x.grad
```

```python
x = torch.tensor([3.0], requires_grad=True) # x = 3.0
y = x ** 2 # y = x ^ 2
y.backward() # y' = 2 x
print(x.grad) # y' = 2 * 3.0 = 6
```

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()

        self.hidden = nn.Linear(1, 4)
        self.relu = nn.ReLU()
        self.output = nn.Linear(4, 1)

    def forward(self, x):

        x = self.hidden(x)
        x = self.relu(x)
        x = self.output(x)

        return x

net = MLP()
print(net)
```

```python
parm = list(net.parameters())

print(parm)
print(len(parm))
```

```python
data = torch.rand(1)
output = net(data)

print(output)
```

```python
target = torch.randn(10)
target = target.view(1, -1)

mse = nn.MSELoss()
result = mse(output, target)

print(result)
```
